```python

# STEP 1: Load and inspect the combined LASI dataset

import pandas as pd
import numpy as np
from google.colab import files
from pathlib import Path

# 1. Upload the LASI combined Excel file
uploaded = files.upload()

excel_files = [
    name for name in uploaded
    if name.lower().endswith((".xlsx", ".xls"))
]

if not excel_files:
    raise ValueError("Please upload the combined LASI Excel workbook.")

# Select the LASI workbook if multiple files were uploaded
lasi_file = next(
    (name for name in excel_files if "lasi" in name.lower()),
    excel_files[0]
)

# 2. Load the main worksheet
xls = pd.ExcelFile(lasi_file)
print("Workbook:", lasi_file)
print("Available sheets:", xls.sheet_names)

df_lasi_original = pd.read_excel(lasi_file, sheet_name=0)

# Preserve a separate working copy
df_lasi = df_lasi_original.copy()

# Remove gender from the working copy if present
gender_cols = [
    c for c in df_lasi.columns
    if c.strip().lower() in {"gender", "sex"}
]
df_lasi = df_lasi.drop(columns=gender_cols)

print("\nOriginal shape:", df_lasi_original.shape)
print("Working shape:", df_lasi.shape)
print("Gender/sex columns removed:", gender_cols or "None found")

# 3. Inspect structure
print("\n--- COLUMN NAMES ---")
print(df_lasi.columns.tolist())

print("\n--- DATA TYPES ---")
print(df_lasi.dtypes.to_string())

print("\n--- FIRST FIVE ROWS ---")
display(df_lasi.head())

print("\n--- MISSING VALUES ---")
missing = pd.DataFrame({
    "missing_count": df_lasi.isna().sum(),
    "missing_percent": (
        df_lasi.isna().mean() * 100
    ).round(2)
})
display(missing.sort_values("missing_percent", ascending=False))

# 4. Duplicate checks
print("\n--- DUPLICATE CHECKS ---")
print("Exact duplicate rows:", df_lasi.duplicated().sum())

for col in ["person_id", "household_id"]:
    if col in df_lasi.columns:
        print(
            f"Rows with repeated {col}:",
            df_lasi[col].duplicated(keep=False).sum()
        )
        print(
            f"Unique {col} values:",
            df_lasi[col].nunique(dropna=True)
        )

# 5. Inspect unique values and possible survey codes
print("\n--- UNIQUE VALUES / CATEGORY CHECK ---")
for col in df_lasi.columns:
    s = df_lasi[col]
    print(f"\n{col}:")
    print("Unique non-null values:", s.nunique(dropna=True))
    if s.nunique(dropna=True) <= 20:
        print(s.value_counts(dropna=False).to_dict())
    else:
        print("Examples:", s.dropna().unique()[:10].tolist())

# 6. Numeric ranges to flag implausible values
print("\n--- NUMERIC SUMMARY ---")
display(df_lasi.describe(include="all").T)

# 7. Check key fields without assuming their definitions
key_fields = [
    "current_age",
    "last_menstrual_period_year",
    "hysterectomy_excessive_bleeding",
    "spatial_orientation_score",
    "heart_disease",
    "avg_systolic_bp",
    "avg_diastolic_bp",
    "bmi",
    "state"
]

print("\n--- KEY FIELD INSPECTION ---")
for col in key_fields:
    if col in df_lasi.columns:
        print(f"\n{col}")
        print("Non-null:", df_lasi[col].notna().sum())
        print("Distinct values:", df_lasi[col].nunique(dropna=True))
        print("Examples:", df_lasi[col].dropna().unique()[:15].tolist())

print("\nStep 1 inspection complete.")
```

```python

# STEP 2: Audit repeated respondents and menopause-target feasibility

import pandas as pd
import numpy as np

# Check that Step 1 has been run
assert "df_lasi" in globals(), "Run Step 1 first."

df_audit = df_lasi.copy()

# A. Investigate repeated person IDs before removing anything
print("A. PERSON ID AUDIT")
print("Rows:", len(df_audit))
print("Unique people:", df_audit["person_id"].nunique())
print("Repeated-ID rows:",
      df_audit["person_id"].duplicated(keep=False).sum())

id_counts = df_audit["person_id"].value_counts()
repeated_ids = id_counts[id_counts > 1].index

duplicate_id_rows = df_audit[
    df_audit["person_id"].isin(repeated_ids)
].sort_values("person_id")

print("\nRows for the first 10 repeated IDs:")
display(duplicate_id_rows.head(10))

# How many repeated IDs have conflicting values?
compare_cols = [
    c for c in df_audit.columns
    if c not in ["person_id", "household_id"]
]

conflict_counts = (
    duplicate_id_rows.groupby("person_id")[compare_cols]
    .nunique(dropna=True)
)

conflicting_ids = (conflict_counts > 1).any(axis=1)

print("\nRepeated IDs with conflicting non-ID data:",
      int(conflicting_ids.sum()))
print("Repeated IDs with identical observed non-ID data:",
      int((~conflicting_ids).sum()))

# B. Check exact duplicate rows
print("\nB. EXACT DUPLICATES")
print("Exact duplicate rows:", df_audit.duplicated().sum())

# C. Audit the last menstrual period year
print("\nC. MENSTRUAL YEAR AUDIT")
lmp = pd.to_numeric(
    df_audit["last_menstrual_period_year"], errors="coerce"
)
age = pd.to_numeric(df_audit["current_age"], errors="coerce")

# LASI Wave 1 was conducted mainly in 2017–2018.
# Use this only as a plausibility screen, not as the target yet.
print("Year counts:")
display(lmp.value_counts(dropna=False).sort_index())

print("\nRows with year after 2018:", int((lmp > 2018).sum()))
print("Rows with year before 1930:", int((lmp < 1930).sum()))
print("Rows with LMP year later than respondent birth-year estimate:",
      int(((lmp > (2018 - age)) & lmp.notna() & age.notna()).sum()))

# Calculate a rough interval for investigation only.
# This is NOT a validated menopause-age target.
rough_years_since_lmp = 2018 - lmp
rough_age_at_lmp = age - rough_years_since_lmp

print("\nApproximate age at last period (diagnostic only):")
display(rough_age_at_lmp.describe())

print("\nApproximate ages outside 25–65:",
      int((~rough_age_at_lmp.between(25, 65)).sum()))

# D. Inspect hysterectomy field and its relationship with LMP availability
print("\nD. HYSTERECTOMY FIELD")
display(
    df_audit["hysterectomy_excessive_bleeding"]
    .value_counts(dropna=False)
    .rename_axis("value")
    .to_frame("count")
)

print("\nLMP availability by hysterectomy field value:")
display(pd.crosstab(
    df_audit["hysterectomy_excessive_bleeding"],
    df_audit["last_menstrual_period_year"].isna(),
    margins=True
))

# E. Cognitive score and cardiovascular indicator distributions
print("\nE. HEALTH INDICATORS")
for col in [
    "spatial_orientation_score",
    "heart_disease",
    "avg_systolic_bp",
    "avg_diastolic_bp"
]:
    print(f"\n{col}")
    print(df_audit[col].describe())
    print(df_audit[col].value_counts(dropna=False).head(15))

print("\nAudit complete. No rows have been deleted.")
```

```python

# STEP 3: Investigate repeated respondent IDs
import pandas as pd

assert "df_lasi" in globals(), "Run Step 1 first."

df_check = df_lasi.copy()

# 1. How many times does each respondent ID appear?
id_counts = df_check["person_id"].value_counts()

print("A. RESPONDENT ID FREQUENCY")
print("IDs appearing once:", int((id_counts == 1).sum()))
print("IDs appearing twice:", int((id_counts == 2).sum()))
print("IDs appearing 3+ times:", int((id_counts >= 3).sum()))
print("Maximum rows for one ID:", int(id_counts.max()))

# 2. Identify columns that conflict within repeated IDs
repeated = df_check[
    df_check["person_id"].isin(id_counts[id_counts > 1].index)
].copy()

compare_cols = [
    c for c in df_check.columns
    if c not in ["person_id", "household_id"]
]

conflicts = repeated.groupby("person_id")[compare_cols].nunique(
    dropna=True
)

conflicting_ids = conflicts.index[(conflicts > 1).any(axis=1)]
conflict_columns = (conflicts > 1).sum().sort_values(ascending=False)

print("\nB. COLUMNS MOST OFTEN CONFLICTING WITHIN THE SAME ID")
display(conflict_columns[conflict_columns > 0].to_frame(
    "number_of_ID_groups_with_conflicts"
).head(15))

print("\nC. EXAMPLES OF CONFLICTING RESPONDENT RECORDS")
example_ids = list(conflicting_ids[:3])

if example_ids:
    display(
        df_check[df_check["person_id"].isin(example_ids)]
        .sort_values("person_id")
    )
else:
    print("No conflicting repeated IDs found.")

# 3. Check whether exact duplicates account for the repeated rows
print("\nD. EXACT DUPLICATE CHECK")
print("Exact duplicate rows:", int(df_check.duplicated().sum()))
print("Rows remaining if exact duplicates alone were removed:",
      len(df_check.drop_duplicates()))

print("\nInvestigation complete. Original df_lasi remains unchanged.")
```

```python

# STEP 4: Remove exact duplicate rows only
assert "df_lasi" in globals(), "Run Step 1 first."

df_lasi_work = df_lasi.copy()

before = len(df_lasi_work)
df_lasi_work = df_lasi_work.drop_duplicates().copy()
after = len(df_lasi_work)

print("Original rows:", before)
print("Exact duplicate rows removed:", before - after)
print("Rows in working copy:", after)
print("Unique person IDs:", df_lasi_work["person_id"].nunique())
print("Remaining rows belonging to repeated IDs:",
      int(df_lasi_work["person_id"].duplicated(keep=False).sum()))

print("\nOriginal df_lasi unchanged:", len(df_lasi) == before)
```

```python

# STEP 5: Profile remaining repeated-ID conflicts
assert "df_lasi_work" in globals(), "Run Step 4 first."

df = df_lasi_work.copy()

id_counts = df["person_id"].value_counts()
repeated_ids = id_counts[id_counts > 1].index

repeated = df[df["person_id"].isin(repeated_ids)].copy()

compare_cols = [
    c for c in df.columns
    if c not in ["person_id", "household_id"]
]

conflicts = repeated.groupby("person_id")[compare_cols].nunique(
    dropna=True
)

conflict_summary = (
    (conflicts > 1)
    .sum()
    .sort_values(ascending=False)
    .rename("IDs_with_conflicts")
    .to_frame()
)

print("Repeated IDs:", len(repeated_ids))
print("\nColumns with conflicting values:")
display(conflict_summary[conflict_summary["IDs_with_conflicts"] > 0])

print("\nExample: repeated IDs with conflicting BMI")
bmi_conflict_ids = conflicts.index[conflicts["bmi"] > 1][:5]

if len(bmi_conflict_ids):
    display(
        df[df["person_id"].isin(bmi_conflict_ids)]
        .sort_values("person_id")
    )
else:
    print("No BMI conflicts found.")
```

```python

# STEP 6: Display genuine BMI conflicts side by side
assert "df_lasi_work" in globals(), "Run Step 4 first."

df = df_lasi_work.copy()

bmi_numeric = pd.to_numeric(df["bmi"], errors="coerce")
df["_bmi_check"] = bmi_numeric

bmi_counts = df.groupby("person_id")["_bmi_check"].nunique(
    dropna=True
)
bmi_conflict_ids = bmi_counts[bmi_counts > 1].index[:5]

cols = [
    "person_id", "household_id", "current_age",
    "height_cm", "weight_kg", "bmi",
    "spatial_orientation_score", "state"
]

print("Number of IDs with conflicting BMI:", int((bmi_counts > 1).sum()))

if len(bmi_conflict_ids):
    display(
        df[df["person_id"].isin(bmi_conflict_ids)][cols]
        .sort_values("person_id")
    )
else:
    print("No conflicting BMI values found.")

df.drop(columns="_bmi_check", inplace=True)
```

```python

# STEP 7: Assess a provisional menopause age-span target
import pandas as pd
import numpy as np

assert "df_lasi_work" in globals(), "Run Step 4 first."

df_model = df_lasi_work.copy()

age = pd.to_numeric(df_model["current_age"], errors="coerce")
lmp_year = pd.to_numeric(
    df_model["last_menstrual_period_year"], errors="coerce"
)

# Approximate age at last period, assuming survey reference year 2018
df_model["approx_age_at_last_period"] = age - (2018 - lmp_year)

# Keep the original values; create labels only for assessment
valid = df_model["approx_age_at_last_period"].between(25, 65)

bins = [25, 30, 35, 40, 45, 50, 55, 60, 66]
labels = ["25–29", "30–34", "35–39", "40–44",
          "45–49", "50–54", "55–59", "60–65"]

df_model["provisional_age_span"] = pd.cut(
    df_model["approx_age_at_last_period"],
    bins=bins,
    labels=labels,
    right=False
)

print("Total rows:", len(df_model))
print("Rows with approximate ages in 25–65:", int(valid.sum()))
print("Rows outside that range or missing:",
      int((~valid).sum()))

print("\nProvisional age-span counts:")
display(
    df_model.loc[valid, "provisional_age_span"]
    .value_counts(sort=False)
    .rename_axis("age_span")
    .to_frame("respondent_count")
)

print("\nThese are provisional bands based on reported last-period year.")
print("They are not yet verified natural-menopause labels.")
```

```python

# STEP 8: Create the first modelling dataset
import pandas as pd
import numpy as np

assert "df_model" in globals(), "Run Step 7 first."

# Keep only rows with a provisional target in the 25–65 range
df_train = df_model[
    df_model["provisional_age_span"].notna()
].copy()

# Features currently available in the LASI dataset
candidate_features = [
    "current_age",
    "age_first_marriage",
    "hot_flashes",
    "insomnia",
    "severe_fatigue",
    "depression_2plus_weeks",
    "ever_alcohol",
    "vigorous_activity_frequency",
    "moderate_activity_frequency",
    "heart_disease",
    "avg_systolic_bp",
    "avg_diastolic_bp",
    "avg_pulse",
    "height_cm",
    "weight_kg",
    "waist_circumference_cm",
    "hip_circumference_cm",
    "bmi",
    "state"
]

features = [c for c in candidate_features if c in df_train.columns]

print("Rows available for modelling:", len(df_train))
print("Target: provisional_age_span")
print("\nTarget counts:")
display(df_train["provisional_age_span"].value_counts(sort=False))

print("\nFeature missingness (%):")
display(
    (df_train[features].isna().mean() * 100)
    .sort_values(ascending=False)
    .round(2)
    .to_frame("missing_percent")
)

print("\nFeature columns available:", len(features))
print("Modelling dataset created. No model has been trained yet.")
```

```python

# STEP 9: Define leakage-safe features and target
from sklearn.model_selection import train_test_split

assert "df_train" in globals(), "Run Step 8 first."

# Exclude current_age because it was used to construct the target.
# Exclude last_menstrual_period_year because it directly defines the target.
# Exclude BMI because it is derived from height and weight.
exclude_cols = {
    "person_id",
    "household_id",
    "current_age",
    "last_menstrual_period_year",
    "approx_age_at_last_period",
    "provisional_age_span",
    "bmi"
}

feature_cols = [
    c for c in df_train.columns
    if c not in exclude_cols
    and c not in ["spatial_orientation_score"]
]

# Use only the available candidate predictors from Step 8
candidate_features = [
    "age_first_marriage",
    "hot_flashes",
    "insomnia",
    "severe_fatigue",
    "depression_2plus_weeks",
    "ever_alcohol",
    "vigorous_activity_frequency",
    "moderate_activity_frequency",
    "heart_disease",
    "avg_systolic_bp",
    "avg_diastolic_bp",
    "avg_pulse",
    "height_cm",
    "weight_kg",
    "waist_circumference_cm",
    "hip_circumference_cm",
    "state"
]

feature_cols = [c for c in candidate_features if c in df_train.columns]

X = df_train[feature_cols].copy()
y = df_train["provisional_age_span"].astype(str)

print("X shape:", X.shape)
print("Target shape:", y.shape)
print("Features:", feature_cols)
print("\nTarget classes:")
print(y.value_counts())

print("\nTarget leakage columns excluded:",
      ["current_age", "last_menstrual_period_year",
       "approx_age_at_last_period", "provisional_age_span"])
```

```python

# STEP 10: Stratified train/test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))

print("\nTraining class distribution (%):")
display(
    (y_train.value_counts(normalize=True).sort_index() * 100)
    .round(2)
    .to_frame("percent")
)

print("\nTesting class distribution (%):")
display(
    (y_test.value_counts(normalize=True).sort_index() * 100)
    .round(2)
    .to_frame("percent")
)

print("\nMissing values in training features:", int(X_train.isna().sum().sum()))
print("Missing values in testing features:", int(X_test.isna().sum().sum()))
```

```python

import pandas as pd
from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score
)
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier
)
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

# --------------------------------------------------
# 1. Encode target labels
# --------------------------------------------------
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

sample_weights = compute_sample_weight(
    class_weight="balanced",
    y=y_train_encoded
)

# --------------------------------------------------
# 2. Define baseline models and boosting variants
# --------------------------------------------------
experiments = [
    (
        "Logistic Regression",
        LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=42
        ),
        False
    ),
    (
        "Decision Tree",
        DecisionTreeClassifier(
            class_weight="balanced",
            random_state=42
        ),
        False
    ),
    (
        "Random Forest",
        RandomForestClassifier(
            n_estimators=300,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        ),
        False
    ),
    (
        "Extra Trees",
        ExtraTreesClassifier(
            n_estimators=300,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        ),
        False
    )
]

learning_rates = [0.03, 0.05, 0.1]

for lr in learning_rates:
    experiments.append((
        f"Gradient Boosting (lr={lr})",
        GradientBoostingClassifier(
            n_estimators=100,
            learning_rate=lr,
            max_depth=3,
            random_state=42
        ),
        True
    ))

for lr in learning_rates:
    experiments.append((
        f"XGBoost (lr={lr})",
        XGBClassifier(
            n_estimators=300,
            max_depth=5,
            learning_rate=lr,
            subsample=0.8,
            colsample_bytree=0.8,
            objective="multi:softprob",
            eval_metric="mlogloss",
            tree_method="hist",
            random_state=42,
            n_jobs=-1
        ),
        True
    ))

for lr in learning_rates:
    experiments.append((
        f"CatBoost (lr={lr})",
        CatBoostClassifier(
            iterations=300,
            learning_rate=lr,
            depth=6,
            loss_function="MultiClass",
            verbose=0,
            random_seed=42,
            allow_writing_files=False
        ),
        True
    ))

# --------------------------------------------------
# 3. Train and evaluate every configuration
# --------------------------------------------------
results = []
fitted_models = {}

for i, (name, model, use_weights) in enumerate(experiments, start=1):
    print(f"[{i}/{len(experiments)}] Training {name} ...")

    pipeline = Pipeline([
        ("preprocessor", clone(preprocessor)),
        ("classifier", model)
    ])

    fit_params = {}
    if use_weights:
        fit_params["classifier__sample_weight"] = sample_weights

    pipeline.fit(
        X_train,
        y_train_encoded,
        **fit_params
    )

    predictions = pipeline.predict(X_test).astype(int)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(
            y_test_encoded, predictions
        ),
        "Balanced Accuracy": balanced_accuracy_score(
            y_test_encoded, predictions
        ),
        "Macro F1": f1_score(
            y_test_encoded,
            predictions,
            average="macro",
            zero_division=0
        )
    })

    fitted_models[name] = pipeline

# --------------------------------------------------
# 4. Rank results
# --------------------------------------------------
results_df = (
    pd.DataFrame(results)
    .sort_values(
        ["Macro F1", "Balanced Accuracy"],
        ascending=False
    )
    .reset_index(drop=True)
)

display(results_df.round(4))
```
